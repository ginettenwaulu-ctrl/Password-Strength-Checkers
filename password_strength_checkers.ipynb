**Project type:** Beginner Python / Cybersecurity

## Project Overview
This project evaluates sample passwords against five basic characteristics: minimum length, uppercase letters, lowercase letters, numbers, and special characters. It displays a checklist score out of five.

## Features
- Checks whether a sample password has at least 12 characters.
- Checks for uppercase and lowercase letters.
- Checks for numbers and special characters.
- Calculates a score out of five.
- Displays a simple result label.

## Skills Practiced
- Python variables and functions
- Conditional statements
- Boolean logic
- Loops and character checks
- Basic cybersecurity concepts
- Testing and documentation

## Tools
Python, Google Colab, and GitHub.

## Limitations
This is an educational checklist, not a complete password-security assessment. It does not check password reuse, breached passwords, predictability, or whether a password is easy to guess. Use fictional examples only.

```
import string

def check_password(password):

    # Check the five password rules
    has_length = len(password) >= 12
    has_uppercase = any(char.isupper() for char in password)
    has_lowercase = any(char.islower() for char in password)
    has_number = any(char.isdigit() for char in password)
    has_special = any(char in string.punctuation for char in password)

    # Calculate the score
    score = sum([
        has_length,
        has_uppercase,
        has_lowercase,
        has_number,
        has_special
    ])

    # Display the results
    print("\nPASSWORD CHECK RESULTS")
    print("----------------------")
    print("At least 12 characters:", has_length)
    print("Contains uppercase:", has_uppercase)
    print("Contains lowercase:", has_lowercase)
    print("Contains a number:", has_number)
    print("Contains a special character:", has_special)

    print("\nScore:", score, "out of 5")

    if score <= 2:
        print("Result: Needs improvement")
    elif score <= 4:
        print("Result: Moderate checklist score")
    else:
        print("Result: Meets all five checklist rules")

    print("\nNote: This is an educational checklist,")
    print("not a complete password security assessment.")
```

```
check_password("BlueSky!2026Example")
```

```
check_password("hello")
```

```
check_password("BlueSky!2026Example")
check_password("hello")
check_password("helloworld123!")
```
